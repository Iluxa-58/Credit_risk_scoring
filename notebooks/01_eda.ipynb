## Вывод по TARGET

Целевой класс является несбалансированным: заявки с TARGET = 1 встречаются значительно реже.

Accuracy не подходит как единственная метрика. Модель, всегда предсказывающая TARGET = 0, может получить высокую accuracy, но не обнаружить ни одного рискованного клиента.

Для оценки моделей потребуется использовать ROC-AUC, PR-AUC, precision, recall и метрики качества вероятностей.

```python
missing = pd.DataFrame({
    "missing_count": df.isna().sum(),
    "missing_percent": df.isna().mean() * 100,
})
missing = missing.sort_values("missing_percent", ascending=False)
missing.head(20)
```

```python
def category_default_rate(data: pd.DataFrame, column: str,) -> pd.DataFrame:
    result = data.groupby(column, dropna=False,)["TARGET"].agg(application="size", default_rate="mean").sort_values("default_rate", ascending=False)
    return result
```

## Проверка data leakage

Для каждого признака необходимо определить, была ли информация доступна в момент подачи кредитной заявки.

SK_ID_CURR является идентификатором и должен быть исключён из модели.

Все операции заполнения пропусков, кодирования категорий, масштабирования и отбора признаков должны обучаться только на тренировочной выборке после разделения данных.
